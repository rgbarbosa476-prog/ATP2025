```python
#tpc2 - Jogo dos Fósforos
import random
print("Olá! Hoje vamos jogar ao Jogo dos Fósforos!")
escolha = int(input("Queres ser o 1º ou o 2º a jogar? "))
print(f"Escolheste ser o {escolha}º a jogar!")
soma = 21
print("O número inicial de fósforos é:", soma)
if escolha == 1:
    while soma > 1:
        pergunta = int(input("Escolhe um número de 1 a 4 para subtrair: "))
        if pergunta in [1, 2, 3, 4] and pergunta < soma:
            print("A tua jogada foi:", pergunta)
            soma -= pergunta
            print("Fósforos restantes:", soma)
            computador = min(5 - pergunta, soma)
            print("A jogada do computador é:", computador)
            soma -= computador
            print("Fósforos restantes:", soma)
            if soma == 1:
                print("Oh não, perdeste a jogada :(")
                break
        else:
            print("Valor não disponível ou demasiado alto.")
    else:
        print("Parabéns! Ganhaste a partida!")
elif escolha == 2:
    computador = random.randint(1, 4)
    print("A jogada do computador é:", computador)
    soma -= computador
    print("Fósforos restantes:", soma)
    while soma > 1:
        pergunta = int(input("Escolhe um número de 1 a 4 para subtrair: "))
        if pergunta in [1, 2, 3, 4] and pergunta < soma:
            print("A tua jogada foi:", pergunta)
            soma -= pergunta
            print("Fósforos restantes:", soma)
            if soma == 1:
                print("Parabéns! Ganhaste a partida!")
                break
            if soma % 5 != 1:
                computador = (soma - 1) % 5
                if computador == 0 or computador > soma:
                    computador = random.randint(1, min(4, soma))
            else:
                computador = random.randint(1, min(4, soma))
            print("A jogada do computador é:", computador)
            soma -= computador
            print("Fósforos restantes:", soma)
            if soma == 1:
                print("Oh não, perdeste a jogada :(")
                break
        else:
            print("Jogada inválida. Tenta novamente.")
else:
    print("Escolha inválida. Reinicia o jogo e escolhe 1 ou 2.")
```
